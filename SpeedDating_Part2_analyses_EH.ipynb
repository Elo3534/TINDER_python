```python
col_demo = df[['genre', 'ethnie']]

# Analyses statistiques des variables démographiques 
results = {}

print("ANALYSES DE LA REPARTITION ET DES FREQUENCES PAR GENRE ET PAR ETHNIE :")

# Définir les palettes complètes
palette_ethnie = {
    'Européen / Caucasien-Américain': '#E41A1C',  # Rouge
    'Asiatique / Pacifique / Asiatique-Américain': '#377EB8',  # Bleu
    'Latino / Hispanique-Américain': '#4DAF4A',  # Vert
    'Autre': '#984EA3',  # Violet
    'Noir / Afro-Américain': '#FF7F00'  # Orange
}

palette_genre = {
    'Homme': "#70c0f9",  # Bleu ciel
    'Femme': '#ff69b4'   # Rose
}

for col in col_demo:
    value_counts = col_demo[col].value_counts()
    labels = value_counts.index
    counts = value_counts.values
    
    print(f"Répartition par {col} :")
    print(value_counts)
    print()

    # Sélectionner la palette appropriée
    if col == 'ethnie':
        palette = palette_ethnie
    else:
        palette = palette_genre
    
    # Créer la liste de couleurs en gérant les labels manquants
    colors_list = []
    for label in labels:
        if label in palette:
            colors_list.append(palette[label])
        else:
            # Couleur par défaut si le label n'est pas dans le dictionnaire
            colors_list.append('#CCCCCC')  # Gris par défaut
            print(f"Attention: label '{label}' non trouvé dans la palette")

    # fonction pour afficher les pourcentages et les fréuences
    def make_autopct(values):
        def my_autopct(pct):
            total = sum(values)
            val = int(round(pct*total/100.0))
            return '{p:.1f}% \n({v:d})'.format(p=pct,v=val)
        return my_autopct

    # Graphique (camembert)
    patches, texts, autotexts = plt.pie(counts, 
                                        autopct=make_autopct(counts), 
                                        shadow=True, 
                                        startangle=90, 
                                        colors=colors_list,
                                        labels=labels  # Ajout des labels directement dans pie()
                                    )
    plt.title(f'Répartition par {col}')
    plt.subplots_adjust(top=1.25)
    plt.axis('equal')  # Assure que le pie chart est circulaire
    plt.legend(patches, labels, loc="lower center", bbox_to_anchor=(0.5, -0.2))
    plt.show()
```

## **6B. Quels sont les attributs les moins désirables chez un partenaire masculin ? Est-ce différent chez les partenaires féminines ?**

Nous allons analyser les résultats des évaluations faites par le participant à l'issue du date, c'est-à-dire à Tf.
Les variables correspondants à ces notes sont : 
    'evaluation_par_participant' = ['attr_eval', 'sinc_eval', 'intel_eval', 'fun_eval', 'amb_eval', 'shar_eval']

Pour que les notes nous révèlent l'importance des attributs au yeux du participant, il faut s'intéresser aux évaluations des participants ayant choisi 'oui' à la question 'decision' qui leur est posée également à l'issue du date (Tf).

Ensuite, pour distinguer des tendances liées au genre, les analyses porteront sur les résultats obtenus lorsque le participant est une femme, et distinctement, lorsque le participant est un homme.

```python
# Filtre sur les données : participants ayant répondu 'yes' à la question 'decision' (décision de revoir le partenaire après le speed date)
df_oui = df[df['decision'] == 'Oui']

# Sélection des attributs 
evaluation_par_participant = ['attr_eval', 'sinc_eval', 'intel_eval', 'fun_eval', 'amb_eval', 'shar_eval']

# Noms des attributs
attributs_noms = ['Attractivité', 'Sincérité', 'Intelligence', 'Amusant', 'Ambitieux', 'Intérêts communs']

print("ANALYSE DES ATTRIBUTS POUR LES PARTICIPANTS AYANT CHOISI 'OUI' À DECISION :")
print("-" * 60)

# Calcul des notes moyennes obtenues par attribut et par genre
moyennes_par_genre = df_oui.groupby('genre')[evaluation_par_participant].mean()

print("Moyennes des notes par attribut et par genre :")
print(moyennes_par_genre.round(2))

# Extraction des données
moyennes_hommes = moyennes_par_genre.loc['Homme']
moyennes_femmes = moyennes_par_genre.loc['Femme']

print("\nCLASSEMENT DES ATTRIBUTS PAR IMPORTANCE (ordre décroissant) :")
print("=" * 70)

# Classement pour les hommes
print("\nPour les HOMMES :")
print("-" * 40)
classement_hommes = moyennes_hommes.sort_values(ascending=False)
for i, (attr, score) in enumerate(classement_hommes.items(), 1):
    nom_attr = attributs_noms[evaluation_par_participant.index(attr)]
    print(f"  {i}. {nom_attr}: {score:.2f}/100")

# Classement pour les femmes
print("\nPour les FEMMES :")
print("-" * 40)
classement_femmes = moyennes_femmes.sort_values(ascending=False)
for i, (attr, score) in enumerate(classement_femmes.items(), 1):
    nom_attr = attributs_noms[evaluation_par_participant.index(attr)]
    print(f"  {i}. {nom_attr}: {score:.2f}/100")
```

```python
# Il est très probable que les variances ne soient pas parfaitement égales puisque nous sommes dans 
# le cas de données réelles. 
# Déjà, parce que nos deux groupes n'ont pas les mêmes effectifs, les estimations de variance vont en
# être directement affectées, l'hypthèse d'égalité des variances n'est donc pas réaliste.
# D'autre part, lors d'expérimentation sur l'Humain, il est souvent constaté une différence 
# de comportement entre les notations ou évaluations des hommes et des femmes,
# Cela se traduit statistiquement par des variances inégales.

# C'est pour cette raison que nous allons utiliser le test t de Welch pour la comparaison des notes données aux attributs.

# En effet, Le test t de Welch est une adaptation du test t de Student. 
# Il est utilisé dans les études médicales lorsque l'hypothèse de l'égalité des variances dans le test t de Student n'est pas satisfaite. 
# Il est utilisé pour comparer la distribution dans deux groupes de patients indépendants.
# Ce qui correspond parfaitement à notre cas d'étude.

# Conditions à remplir pour le test t de Welch :
# 1. Les données doivent être quantitatives continues (c'est le cas ici avec des notes de 0 à 100).
# 2. Les observations doivent être indépendantes (c'est le cas ici avec des participants différents).
# 3. Les distributions des deux groupes doivent être approximativement normales 
# (ce qui est notre cas grâce à la taille de notre échantillon (suffisamment grand), grâce au théorème central limite).
# 4. Pas d'hypothèse d'égalité des variances (comme expliqué au dessus).
```

```python
print("TEST DE SIGNIFICATIVITÉ SUR LES DIFFÉRENCES DE NOTATION ENTRE HOMMES ET FEMMES")
print("-" * 50)

print("\nTests t de Welch (adaptation du test t de Student) pour comparer les moyennes Hommes vs Femmes :")
print("-" * 50)
print(f"{'Attribut':<20} {'p-value':<10} {'Significatif':<12} {'Différence':<12}")
print("-" * 50)

for i, attr in enumerate(evaluation_par_participant):
    # Données pour chaque genre
    hommes_data = df_oui[df_oui['genre'] == 'Homme'][attr]
    femmes_data = df_oui[df_oui['genre'] == 'Femme'][attr]
    
    # Test t de Student
    t_stat, p_value = ttest_ind(hommes_data, femmes_data, equal_var=False)  # Welch's t-test
    
    # Déterminer la significativité
    significatif = "OUI" if p_value < 0.05 else "NON"
    
    # Calculer la différence moyenne
    diff = moyennes_femmes[attr] - moyennes_hommes[attr]

    print(f"{attributs_noms[i]:<20} {p_value:<10.4f} {significatif:<12} {diff:>+8.2f}")
```

Différences statistiquement significatives (p < 0.05) :
  - Attractivité (p = 0.00) : 0.95 points de plus chez les notes attribuées par les hommes.
  - Ambitieux (p = 0.00) : 0.58 points de plus chez les notes attribuées par les femmes.
  - Intelligence (p = 0.00) : 0.49 points de plus chez les notes attribuées par les femmes.


```python
# Histogramme des valeurs moyennes des attributs par genre
plt.figure(figsize=(10, 8))

# Séparation les données par genre et calcul des notes moyennes obtenues par attribut et par genre
moyennes_hommes = df_oui[df_oui['genre'] == 'Homme'][evaluation_par_participant].mean()
moyennes_femmes = df_oui[df_oui['genre'] == 'Femme'][evaluation_par_participant].mean()

# Noms des attributs
attributs_noms = ['Attractivité', 'Sincérité', 'Intelligence', 'Amusant', 'Ambitieux', 'Intérêts communs']

# Position des barres
x = np.arange(len(attributs_noms))
largeur_barre = 0.35

# Créer l'histogramme double
plt.bar(x - largeur_barre/2, moyennes_hommes, largeur_barre, color='#70c0f9', edgecolor='black', alpha=0.7, label='Hommes')
plt.bar(x + largeur_barre/2, moyennes_femmes, largeur_barre, color='#ff69b4', edgecolor='black', alpha=0.7, label='Femmes')

# Ajouter les valeurs sur les barres
for i in range(len(attributs_noms)):
    plt.text(i - largeur_barre/2, moyennes_hommes.iloc[i] + 0.2, f'{moyennes_hommes.iloc[i]:.1f}', 
             ha='center', va='bottom', fontsize=9, fontweight='bold')
    plt.text(i + largeur_barre/2, moyennes_femmes.iloc[i] + 0.2, f'{moyennes_femmes.iloc[i]:.1f}', 
             ha='center', va='bottom', fontsize=9, fontweight='bold')

# Ajouter les annotations de significativité statistique de manière automatisée
significatif_y = 21  # Position verticale pour les annotations

for i, attr in enumerate(evaluation_par_participant):
    # Données pour chaque genre
    hommes_data = df_oui[df_oui['genre'] == 'Homme'][attr]
    femmes_data = df_oui[df_oui['genre'] == 'Femme'][attr]
    
    # Test t de Student
    t_stat, p_value = ttest_ind(hommes_data, femmes_data, equal_var=False)
    
    # Si significatif, ajouter l'annotation
    if p_value < 0.05:
        # Ajouter étoiles de significativité
        plt.text(i, significatif_y, '***', ha='center', va='bottom', 
                fontsize=14, fontweight='bold', color='red')
        
        # Calculer la différence et déterminer la direction
        diff = moyennes_femmes[attr] - moyennes_hommes[attr]
        if diff > 0:
            texte = f'Femmes +{abs(diff):.2f}'
        else:
            texte = f'Hommes +{abs(diff):.2f}'
            
        # Ajouter la différence
        plt.text(i, significatif_y - 1, texte, ha='center', va='bottom', 
                fontsize=8, fontweight='bold', color='red', style='italic')

plt.title('Distribution des moyennes obtenues par attributs\n*** Différences statistiquement significatives (p < 0.05)', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Attributs', fontsize=12, fontweight='bold')
plt.ylabel('Note moyenne obtenue /100', fontsize=12, fontweight='bold')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.xticks(x, attributs_noms, rotation=45, ha='right')
plt.ylim(0, 23)

plt.tight_layout()
plt.show()
```

```python
# Nous allons chercher si les participants accordent de l'importance à être de la même ethnie que leur partenaire potentiel,
# en cherchant une corrélation entre same_ethnie et decision 'oui'.

# Table de contingence
contingence_table = pd.crosstab(df['same_ethnie'], df['decision'])

# Heatmap de la table de contingence
plt.figure(figsize=(8, 6))
sns.heatmap(contingence_table, annot=True, fmt='d', cmap='Blues')
plt.title('Décision de revoir le partenaire x Appartenance à la même ethnie \nà l\'issue du date')
plt.xlabel('même ethnie entre le participant et le partenaire')
plt.ylabel('décision de revoir le partenaire de date')

plt.show()

print('-'*50)
print("\nPourcentages par ligne :")
print(pd.crosstab(df['same_ethnie'], df['decision'], normalize='index') * 100)
```

```python
# Test Chi² complet
chi2, p_value, dof, expected = chi2_contingency(contingence_table)

print(f"\nRÉSULTATS DU TEST CHI² :")
print(f"Statistique Chi² : {chi2:.3f}")
print(f"P-value : {p_value:.3f}")
print(f"Degrés de liberté : {dof}")
print(f"Effectif total : {contingence_table.values.sum():,}".replace(',', ' '))

print(f"\nSEUIL DE SIGNIFICATIVITÉ :")
alpha = 0.05
print(f"Seuil = {alpha}")

print(f"\nCONCLUSION :")
if p_value < alpha:
    print("REJET de l'hypothèse nulle d'indépendance")
    print("Il existe une relation SIGNIFICATIVE entre l'ethnie et la décision")
else:
    print("NON-REJET de l'hypothèse nulle d'indépendance") 
    print("Aucune relation SIGNIFICATIVE entre l'ethnie et la décision")
```

L'hypothèse nulle "La même appartenance ethnique n'a aucune influence sur la décision de revoir le partenaire de date" ne peut pas être rejetée puisque notre seuil alpha (0,05) est inférieur à notre p-value (0,098).

Sur la base de ces données, on ne peut pas affirmer statistiquement que la même appartenance ethnique influence la décision de se revoir. 
Les tendances observées pourraient être le fruit du hasard. Mais regardons-les.

### *Les intérêts communs ont-ils une influence sur la décision de revoir un partenaire de date ?*

Nous voulons savoir si la corrélation moyenne des évaluations d'intérêts du participant et du partenaire est différentes entre les participants ayant choisi de se revoir (variable 'decision', modalité 'Oui') et ceux ayant choisi de ne pas se revoir (variable 'decision', modalité 'Non').

Pour comparer les moyennes d'une variable continue avec une variable catégorielle à deux modalités, il est préférable de faire un Test T.
Nos données le permettent : 
***corr_interest*** : Corrélation entre les évaluations d'intérêts du participant et du partenaire. 
  - Numérique 
  - continue
  - Valeur comprise entre -1 et 1.
***decision*** : Décision du participant à l'issue du date.
  - catégorielle 
  - binaire 
  - valeur = 'Oui' ou 'Non'

Grand échantillon donc le Test T devient robuste à la non-normalité (Théorème Central Limite : même si la distribution de l'échantillon n'est pas normale, la disribution des moyennes le devient au fur et à mesure que n augmente. Effet à partir de n=30 (grâce aux moyennes, les valeurs très basses et très hautes se compensent, la distribution des moyennes est lissée)).




```python
# Séparation des groupes catégoriels
corr_oui = df[df['decision'] == 'Oui']['corr_interest']
corr_non = df[df['decision'] == 'Non']['corr_interest']

# Vérification des données
print("Vérification des données:")
print(f"Type de corr_oui: {type(df[df['decision'] == 'Oui']['corr_interest'])}")
print(f"Valeurs manquantes Oui: {df[df['decision'] == 'Oui']['corr_interest'].isna().sum()}")
print(f"Valeurs manquantes Non: {df[df['decision'] == 'Non']['corr_interest'].isna().sum()}")
```

```python
# Dans la pratique, avec de grands échantillons, on peut faire directement le Test T
print("TEST T")


t_stat, p_value = ttest_ind(corr_oui, corr_non, equal_var=False)
print(f"Test T de Welch: t = {t_stat:.4f}, p = {p_value:.4f}")

# Statistiques descriptives
print(f"\nStatistiques descriptives:")
print(f"Groupe OUI: M = {corr_oui.mean():.4f}, SD = {corr_oui.std():.4f}, n = {len(corr_oui)}")
print(f"Groupe NON: M = {corr_non.mean():.4f}, SD = {corr_non.std():.4f}, n = {len(corr_non)}")
print(f"Différence: {corr_oui.mean() - corr_non.mean():.4f}")

# Interprétation
alpha = 0.05
if p_value < alpha:
    print(f"\nDIFFÉRENCE SIGNIFICATIVE (p < {alpha})")
    direction = "plus élevée" if corr_oui.mean() > corr_non.mean() else "plus faible"
    print(f"La corrélation des intérêts est significativement {direction} dans le groupe OUI")
else:
    print(f"\nPAS DE DIFFÉRENCE SIGNIFICATIVE (p ≥ {alpha})")
```

Les personnes qui décident de se revoir ont tendance à avoir une légèrement plus forte corrélation entre leurs intérêts communs (0,21 contre 0,20).
Mais l'effet observé est trop faible (différence de 0,014 sur une échelle allant de -1 à 1) pour être statistiquement significatif (p_value > 0,05).

Nous conclurons donc que la similarité entre les intérêts du participant et les intérêts du partenaire n'a pas un effet statistiquement significatif sur la décision de se revoir.

**En conclusion sur l'influence de l'origine ethnique et des centres d'intérêts partagés sur la décision de revoir un partenaire de date :**

Au vue des résultats obtenus, il apparaît que ni l'origine ethnique commune ni les intérêts partagés n'exercent d'influence statistiquement significative sur la décision de poursuivre ou non une relation après un premier rendez-vous. 

L'appartenance ethnique commune, bien que montrant des tendances directionnelles modestes dans les données observées, ne présente pas d'effet significatif (p = 0,098), ce qui confirme son faible poids dans les préférences déclarées des participants lors de l'inscription. 
De même, la corrélation entre les intérêts du participant et les intérêts du partenaire, bien que très légèrement plus élevée chez les personnes choisissant de se revoir, ne dépasse pas le seuil de significativité statistique (p = 0,065). 

Ces résultats suggèrent que d'autres facteurs, tels que l'intelligence perçue, la sincérité ressentie, l'attraction physique ou encore le caractère amusant du partenaire, jouent probablement un rôle plus déterminant dans la formation des couples que les critères ethniques ou la similarité des centres d'intérêt. 

L'expérience démontre ainsi que, dans ce contexte de rencontres, les individus privilégient des dimensions relationnelles plus complexes que le simple partage d'origines ou d'intérêts communs.

Il y a un biais dans l'analyse, en effet, les autoévaluations étaient notées avec 100 points répartis sur 5 attributs tandis que l'évaluation du partenaire de date était une répartition de 100 points sur 6 attributs.
Nous allons devoir rééchelonner les évaluations du partenaires.

```python
attributs_communs = ['attr', 'sinc', 'intel', 'fun', 'amb']

# Rééchelonnage des évaluations faites par les partenaires 
# pour que la somme des 5 attributs (attractivité, sincérité, intelligence, amusant et ambitieux) soit égale à 100 :
def rescale_partner_scores(row):
    total_5_attrs = sum(row[f'{attri}_by_partner'] for attri in attributs_communs)
    for attri in attributs_communs:
        row[f'{attri}_by_partner_rescaled'] = (row[f'{attri}_by_partner'] / total_5_attrs) * 100
    return row

df = df.apply(rescale_partner_scores, axis=1)
```

```python
# Analyser corrélation entre autoévaluation du articipant et évaluation du partenaire pour chaque attribut séparément
# une fois avec les autoévaluations faites à l'inscription
# une fois avec les autoévaluations faites le lendemain

# Stocker les résultats dans des listes
ameliorations = []
correlations_init = []
correlations_lend = []
attributs_list = []

print("ÉVOLUTION de l'autoévaluation : comparaison des corrélations entre les autoévaluations (à l'inscription et le lendemain du date)")
print("AVEC l'évaluation du participant par le partenaire :")
print("-" * 50)

for attri in attributs_communs:
    corr_init = df[f'{attri}_autoeval_1'].corr(df[f'{attri}_by_partner_rescaled'])
    corr_lend = df[f'{attri}_autoeval_2'].corr(df[f'{attri}_by_partner_rescaled'])
    amelioration = corr_lend - corr_init
    
    tendance = "✓ AMÉLIORATION" if amelioration > 0 else "✗ DÉTÉRIORATION"
    print(f"{attri}: {corr_init:.3f} → {corr_lend:.3f} ({amelioration:+.3f}) {tendance}")
    
    ameliorations.append((attri, corr_init, corr_lend, amelioration))
    correlations_init.append(corr_init)
    correlations_lend.append(corr_lend)
    attributs_list.append(attri)
```

## **6F. Pour obtenir un deuxième rendez-vous, est-il préférable d'être le premier speed date de la soirée ou le dernier ?**

```python
# Répartition en pourcentages des résultats de match par ordre de passage dans la soirée
print("PROPORTIONS DES RESULTATS DE DATES SELON L\'ORDRE DE PASSAGE :")
resultat_par_ordre = df.groupby('ordre_date')['match'].value_counts(normalize=True).unstack().round(2)
resultat_par_ordre
```

Chaque date est représenté par deux lignes, mais puisqu'il l'est aussi bien pour un date dont 'match'= Non, que pour 'match' = Oui, les proportions restent les mêmes.

```python
# Taux de match par ordre de date
taux_match = resultat_par_ordre['Oui'] * 100


# Représentation graphique
plt.figure(figsize=(12, 6))

# Graphique en barres
ax = taux_match.plot(kind='bar', alpha=0.7, color='skyblue', label='Taux de match')

# Ajouter les valeurs au-dessus des barres
for i, v in enumerate(taux_match):
    plt.text(i, v + 1.5, f'{v:.1f}', 
             ha='center', va='bottom', fontsize=9, fontweight='bold')

# Spline cubique pour une courbe très lisse
x_positions = np.arange(len(taux_match))
cs = CubicSpline(x_positions, taux_match.values)

x_smooth = np.linspace(x_positions.min(), x_positions.max(), 200)
y_smooth = cs(x_smooth)

plt.plot(x_smooth, y_smooth, "r-", linewidth=3, label='Courbe de tendance')

# Formater l'axe des y en pourcentages
#plt.gca().yaxis.set_major_formatter(PercentFormatter(1.0))


plt.title('Taux de match selon l\'ordre du date')
plt.ylabel('Pourcentage de matches')
plt.xlabel('Ordre du date')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

L'analyse de l'histogramme révèle une variation périodique du taux de match en fonction de l'ordre des dates. Cette pseudo-cyclicité suggère l'influence d'un facteur de fatigue ou de saturation cognitive chez l'utilisateur au fil de la soirée.

Notre objectif est de vérifier si l'ordre du date influence significativement la probabilité de match entre le participant et le partenaire.
Nous avons une variable qualitative ordinale ('ordre_date') avec 22 valeurs distinctes et une variable catégorielle binaire ('match') dont les deux modalités sont 'Oui' et 'Non'.

Notre test de Chi² va nous renseigner sur quelle hypothèse est la bonne entre :
H0 : "L'ordre du date et la décision de match sont INDÉPENDANTS"
H1 : "L'ordre du date et la décision de match sont LIÉS"

```python
# Test d'indépendance entre ordre et match
table_contingence = pd.crosstab(df['ordre_date'], df['match'])
chi2, p_value, dof, expected = chi2_contingency(table_contingence)

print(f"Test du Chi-deux: χ² = {chi2:.3f}, p = {p_value:.4f}")

# Test spécifique premier vs dernier
table_premier_dernier = pd.crosstab(
    df['ordre_date'].isin([df['ordre_date'].min(), df['ordre_date'].max()]), 
    df['match']
)
chi2_spec, p_spec, _, _ = chi2_contingency(table_premier_dernier)
print(f"Test premier vs dernier: p = {p_spec:.4f}")
```

Notre p-value est inférieure à notre seuil alpha (0,05) : p= 0,0043.
Cela nous permet de rejeter l'hypothèse H0 disant que l'ordre du date et la décison de match sont indépendants.
De ce fait, nous pouvons conclure que l'ordre du date et la décision de se revoir sont liés.
La distribution des match observée dans l'histogramme ci-dessus n'est donc pas aléatoire et dûe au hasard, elle est influencée par la position du date lors de la soirée.

Le test comparant la position 1 et la position 22 (dernière position) montre une différence très significative puisque la p-value est de 0,0001. Il existe donc une différence significative très marquée entre 23,0 % au premier date et 30,8 % au 22ème date de la soirée.

En regardant de nouveau l'histogramme, nous pouvons dire que les matches sont favorisés lors des dates ayant lieu à la première position (23,0%), à la 9ème position (21,4 %), à la 20ème position(19,1%) mais surtout à la dernière position (22ème position) (30,8 %).

A contrario, passer à la 15ème ou 17ème position (respectivement 12,9 et 12,8 %) n'est pas avantageux pour obtenir un date, et encore moins de passer à la 21ème position (5,0 %).

L'ordre du speed date influence significativement les chances de match, avec un avantage net pour les premières et dernières positions. Les participants devraient stratégiquement viser ces créneaux pour maximiser leurs chances de deuxième rendez-vous.

L'analyse des taux de match en fonction de l'ordre de passage révèle une influence significative de la position dans la séquence, qui peut s'expliquer par des effets de contexte et de fatigue décisionnelle plutôt que par des biais de mémorisation, puisque les évaluations sont recueillies immédiatement après chaque date.


Habituellement dans les expériences faisant intéragir l'attention et la cognition de chaque individu de l'échantillon, il est à noté qu'il existe des effets à prendre en compte dans l'interprétation des résultats :

  - *Effet de primauté* : Les premiers participants bénéficient de l'énergie du début de soirée, de l'excitation de participer à l'expérience et de l'absence de point de comparaison. Le premier date sert souvent de référence pour établir la "norme" de la soirée.

  - *Effet de fatigue cognitive et de saturation (Milieu)* : Au coeur de la séquence, la charge mentale s'accumule. La capacité d'attention et d'enthousiasme diminue, conduisant à une évaluation plus stricte ou moins investie des profils intermédiaires.

  - *Effet de contraste et de "dernière impression" (Récence)* : Le dernier participant capitalise sur un effet de contraste positif si la fatigue du milieu de séquence a créé un creux. Il laisse à la dernière impression immédiate, une position stratégique avant la délibération finale.

Conclusion : La structure même de l'événement, en créant des états cognitifs et émotionnels variables, avantage objectivement les places en début et en fin de séquence. Il est donc stratégique pour un participant de viser ces créneaux pour maximiser sa visibilité et son impact.



La valeur de 5,0 % de matches à la 21ème position semble étonnante, vérifions s'il n'y a pas une explication dans la structure de l'expérience.

Pour exemple, l'ordre n°22 n'existe que dans 1 wave sur 21. Le "dernier date" est donc placé à un ordre différent selon les waves.

L'analyse initiale surestime probablement l'effet 'dernier date' car elle se base sur une seule wave atypique (21) qui a la session la plus longue (22 dates). Une analyse incluant tous les derniers dates réels (quel que soit leur numéro d'ordre) est nécessaire pour conclure de façon rigoureuse.

D'autre part, les effectifs de chaque wave (de 36 à 729 participants) sont différents.
Cela et le nombre fluctuant de dates par wave biaisent l'interprétation de nos analyses.

```python
# Compter le nombre de dates pour l'ordre #22
dates_ordre_22 = df[df['ordre_date'] == 22]
nb_dates_22 = len(dates_ordre_22)
# compter le nombre de matches aux dates placés en 22ème position (à diviser par 2 à cause des double-lignes pour un même date)
nb_matchs_22 = int(((dates_ordre_22['match'] == 'Oui')/2).sum())

print(f"Ordre #22 :")
print(f"  - Waves concernées : {dates_ordre_22['numero_exp'].unique()}")
print(f"  - Nombre total de lignes pour les dates de la position 22 : {nb_dates_22}")
print(f"  - Nombre de dates ayant eu lieu en position n°22 : {nb_dates_22/2:.0f}")
print(f"  - Nombre de matches au 22ème date : {nb_matchs_22}")
print(f"  - Taux de match : {(nb_matchs_22/(nb_dates_22/2))*100:.1f}%")
```

Lors de la wave n°21, seule wave à avoir eu l'ordre n°22 dans l'expérience, il y a eu 13 dates à la position 22 (un seul date est représenté par deux lignes dans nos données, une première ligne avec le retour du participant, une deuxième ligne avec le retour du partenaire).
On voit que 4 d'entre eux ont été des match. C'est bien 30,8 % de match pour cette position dans cette wave. Cependant on prend conscience que le calcul est réalisé sur très peu de répétitions (13). LA taille de notre échantillon devient donc trop petit.


```python
# Taux de match global (référence)
taux_global = (df['match'] == 'Oui').mean()
nb_matchs_global = (df['match'] == 'Oui').sum()
nb_total_global = len(df)

print(f"\nTaux de match global : {taux_global*100:.1f}%")

# Test de proportion (ordre #22 vs population globale)
count = np.array([nb_matchs_22, nb_matchs_global])
nobs = np.array([nb_dates_22, nb_total_global])

z_stat, p_value = proportions_ztest(count, nobs, alternative='larger')
print(f"Test Z de proportion : z = {z_stat:.3f}, p = {p_value:.4f}")

if p_value < 0.05:
    print("✓ Différence STATISTIQUEMENT SIGNIFICATIVE")
else:
    print("✗ Différence NON SIGNIFICATIVE")
```

Les tests statistiques confirment notre intuition.
L'analyse du taux de match à l'ordre de passage n°22 n'est pas statistiquement significatif en comparaison au taux de match global de notre expérience.
L'analyse statistique ne montre aucune preuve que l'ordre #22 améliore les chances de match. 
Au contraire, les données suggèrent plutôt un désavantage (z négatif), mais cette différence n'est pas statistiquement significative (p = 0,5743). L'effet 'dernier date' n'est pas confirmé dans l'ensemble des données.

En résumé : L'ordre n°22 n'apporte ni avantage ni désavantage statistiquement démontrable.

L'ordre de passage influence significativement les chances de succès, mais cette influence doit être nuancée par la structure variable des sessions de l'expérience.

Le premier date bénéficie de l'effet de primauté. Le dernier date capitalise sur l'effet de récence (effet de contraste). À l'inverse, les positions médianes, particulièrement les 15ème et 17ème passages (autour de 13 % de succès), semblent pâtir d'une forme de saturation de l'attention.

Cette analyse comporte toutefois une limite importante : la variabilité des sessions. Alors que certaines soirées ne comptent que 5 dates successifs, d'autres en comportent jusqu'à 22. Donc, si l'effet de primauté est robuste et généralisable à toutes les sessions, l'effet de récence observé sur le dernier date doit être interprété avec prudence car il varie considérablement selon la longueur de la session et repose parfois sur de petits effectifs.

Malgré ces nuances, une recommandation pratique s'impose : les participants ont intérêt à privilégier les créneaux en début de session. Que ce soit pour profiter de la fraîcheur décisionnelle du début ou de l'excitation de début d'expérience, cette positions offre un avantage dans la compétition attentionnelle que constitue le speed dating.
