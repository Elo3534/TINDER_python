# **1. OBJECTIF DE L'ETUDE**

L'équipe marketing a besoin d'aide pour un nouveau projet. Confrontée à une baisse du nombre de correspondances,
elle cherche à comprendre ce qui motive l'intérêt des utilisateurs .

Elle a décidé de mener une expérience de speed dating avec des personnes qui devaient donner à Tinder de
nombreuses informations sur elles. 
Tinder a ensuite collecté les données de cette expérience. Chaque ligne de l'ensemble de données représente un speed
dating entre deux personnes et indique si chacune d'elles a secrètement accepté de se rendre à un second rendez-vous
avec l'autre.  

L'expérience s'est déroulée entre 2002 et 2004.

Au cours de ces rencontres, les participants avaient un premier rendez-vous de quatre minutes avec chaque
participant du sexe opposé. À l'issue de ces quatre minutes, il leur était demandé s'ils souhaitaient revoir leur
partenaire. Ils devaient également évaluer leur partenaire selon six critères : l'attrait, la sincérité, l'intelligence, le plaisir,
l'ambition et les centres d'intérêt communs.
L'ensemble de données comprend également des données de questionnaires recueillies auprès des participants à
différentes étapes du processus. Ces données comprennent : les données démographiques, les habitudes amoureuses,
la perception de soi selon ses principaux attributs, les croyances sur ce que les autres apprécient chez un partenaire et
des informations sur le mode de vie. (Consultez le document « Speed+Dating+Data+Key.doc » du dossier pour plus de détails.)
  
   
   
  
    
  OBJECTIF = **Pourquoi les gens acceptent-ils un deuxième rendez-vous ?**

Pour y répondre :

  + *Quels sont les attributs les moins désirables chez un partenaire masculin ? Est-ce différent chez les partenaires féminines ?*

  + *Dans quelle mesure les gens accordent-ils de l’importance à l’attractivité dans le choix d’un partenaire potentiel par rapport à son impact réel ?*

  + *Les intérêts communs sont-ils plus importants qu’une origine raciale commune ?*

  + *Les gens peuvent-ils prédire avec précision leur propre valeur perçue sur le marché des rencontres ?*

  + *Pour obtenir un deuxième rendez-vous, est-il préférable d'être le premier speed date de la soirée ou le dernier ?*

D'autre part, utiliser la médiane plutôt que la moyenne permet de mieux préserver la distribution des données.

```python
# REMPLACEMENTS DES VALEURS MANQUANTES PAR LA MÉDIANE 

def traitement_100Nan(df_base, df_en_cours, dict_100NAN):
        # Applique le traitement spécifié à plusieurs ensembles de colonnes
    
        # Arguments:
        df_base #: DataFrame pour le calcul des médianes
        df_en_cours #: DataFrame à modifier
        dict_100NAN  # : Dictionnaire {nom_ensemble: [liste_colonnes]}
        
            
        # Création d'une copie de df2 pour ne pas modifier l'original
        df_resultat = df_en_cours.copy()
        dict_medians = {}  # Dictionnaire pour stocker les médianes de chaque ensemble
    
                # 1. Calcul des médianes
        for nom_ensemble, colonnes in dict_100NAN.items():
            medians = df_base[colonnes].median()
            dict_medians[nom_ensemble] = medians
            print(f"\n Médianes pour l'ensemble {nom_ensemble} :\n", medians) 
                # 2. Remplacement des NaN par les médianes
        for nom_ensemble, colonnes in dict_100NAN.items():
            for col in colonnes:
                mask = df_resultat[col].isna()
                df_resultat.loc[mask, col] = dict_medians[nom_ensemble][col]      
             
        return df_resultat

# Application du traitement 
dict_100NAN = {
    'cols1_1': ['attr1_1', 'sinc1_1', 'intel1_1', 'fun1_1', 'amb1_1', 'shar1_1'],
    'cols1_2': ['attr1_2', 'sinc1_2', 'intel1_2', 'fun1_2', 'amb1_2', 'shar1_2'],
    'cols_pf': ['pf_o_att', 'pf_o_sin', 'pf_o_int', 'pf_o_fun', 'pf_o_amb', 'pf_o_sha'],
    'cols2_1': ['attr2_1', 'sinc2_1', 'intel2_1', 'fun2_1', 'amb2_1', 'shar2_1']
}
df= traitement_100Nan(df, df, dict_100NAN)
```

```python
# Nous allons déterminerla méthode de détection des outliers la plus adaptée aux variables ATTRIBUTS en fonction de leur distribution.
# Autrement dit, nous allons appliquer la méthode de l'écart interquartile (IQR) pour les distributions non-normales,
# et la méthode du Z-score pour les distributions normales.

# Cependant, nous allons remplacer les outliers par des valeurs Nan, afin d'éviter qu'elles n'influencent les analyses statistiques.
# Nous pourrons ensuite remplacer ces NaN par la moyenne des k voisins les plus proches
# grâce à l'algorithme K-NN (K-nearest Neighbors)
```

```python
# Détection et traitement des outliers avec imputation K-NN

class OutlierTreatment:
    def __init__(self, alpha=0.05):
        self.alpha = alpha
        self.outliers_info = {}
        self.colonnes_traitees = []
        self.methode_utilisee = {}  # Nouveau: stocker la méthode utilisée par colonne
    
    def detecter_outliers_auto(self, data):
        """Détecte les outliers avec la méthode appropriée et retourne aussi la méthode utilisée"""
        data_array = np.array(data).flatten()
        
        if len(data_array) < 3:
            return np.array([False] * len(data_array)), "Données insuffisantes"
        
        # Choix de la méthode basé sur la normalité
        if len(data_array) <= 5000:
            try:
                _, p_value = shapiro(data_array)
                use_zscore = p_value > self.alpha
                methode = "Z-score" if use_zscore else "IQR"
            except:
                skewness_val = abs(skew(data_array))
                use_zscore = skewness_val < 0.5
                methode = "Z-score" if use_zscore else "IQR"
        else:
            skewness_val = abs(skew(data_array))
            use_zscore = skewness_val < 0.5
            methode = "Z-score" if use_zscore else "IQR"
        
        if use_zscore:
            # Méthode Z-score
            z_scores = np.abs((data_array - np.mean(data_array)) / np.std(data_array))
            return z_scores > 3, f"Z-score (p-value: {p_value:.4f})" if 'p_value' in locals() else f"Z-score (skewness: {skewness_val:.3f})"
        else:
            # Méthode IQR
            Q1 = np.percentile(data_array, 25)
            Q3 = np.percentile(data_array, 75)
            IQR = Q3 - Q1
            lower = Q1 - 1.5 * IQR
            upper = Q3 + 1.5 * IQR
            return (data_array < lower) | (data_array > upper), f"IQR (skewness: {skewness_val:.3f})"
    
    def traiter_outliers_dataset(self, df, colonnes):
        """Traite les outliers en les remplaçant par NaN"""
        df_traite = df.copy()
        self.outliers_info = {}
        self.colonnes_traitees = []
        self.methode_utilisee = {}
        
        print("🔍 DÉTECTION ET TRAITEMENT DES OUTLIERS")
        print("=" * 70)
        
        for colonne in colonnes:
            try:
                data_original = df[colonne].copy()
                outliers_mask, methode = self.detecter_outliers_auto(data_original.dropna())
                n_outliers = outliers_mask.sum()
                
                if n_outliers > 0:
                    # Remplacer les outliers par NaN
                    data_clean = data_original.copy()
                    indices_outliers = data_original.dropna().index[outliers_mask]
                    data_clean.loc[indices_outliers] = np.nan
                    df_traite[colonne] = data_clean
                    
                    # Stocker les informations
                    self.outliers_info[colonne] = {
                        'outliers_detectes': n_outliers,
                        'pourcentage': (n_outliers / len(data_original.dropna())) * 100,
                        'indices': indices_outliers,
                        'methode': methode  # Stocker la méthode utilisée
                    }
                    self.colonnes_traitees.append(colonne)
                    self.methode_utilisee[colonne] = methode
                    
                    print(f"✅ {colonne:20} | {methode:30} | {n_outliers:3} outliers → NaN ({(n_outliers/len(data_original.dropna()))*100:.1f}%)")
                else:
                    print(f"✓   {colonne:20} | {methode:30} | Aucun outlier")
                    
            except Exception as e:
                print(f"❌ {colonne:20} | Erreur: {e}")
        
        return df_traite
    
    def imputer_knn(self, df, n_neighbors=5):
        """Impute les valeurs manquantes avec K-NN"""
        print(f"\n🔧 IMPUTATION K-NN (k={n_neighbors})")
        print("=" * 60)
        
        # Préparer les données
        data_imputation = df[self.colonnes_traitees].copy()
        
        # Standardiser les données pour K-NN
        scaler = StandardScaler()
        data_scaled = scaler.fit_transform(data_imputation)
        
        # Appliquer K-NN imputer
        imputer = KNNImputer(n_neighbors=n_neighbors)
        data_imputed_scaled = imputer.fit_transform(data_scaled)
        
        # Revenir à l'échelle originale
        data_imputed = scaler.inverse_transform(data_imputed_scaled)
        
        # Créer le DataFrame final
        df_final = df.copy()
        for i, colonne in enumerate(self.colonnes_traitees):
            df_final[colonne] = data_imputed[:, i]
        
        # Statistiques de l'imputation
        nan_avant = df[self.colonnes_traitees].isna().sum().sum()
        nan_apres = df_final[self.colonnes_traitees].isna().sum().sum()
        
        print(f"Valeurs manquantes avant K-NN: {nan_avant}")
        print(f"Valeurs manquantes après K-NN: {nan_apres}")
        print(f"Colonnes traitées: {len(self.colonnes_traitees)}")
        
        # Afficher le résumé des méthodes utilisées
        self.afficher_resume_methodes()
        
        return df_final
    
    def afficher_resume_methodes(self):
        """Affiche un résumé des méthodes utilisées"""
        if not self.methode_utilisee:
            return
            
        print(f"\n📊 RÉSUMÉ DES MÉTHODES DE DÉTECTION")
        print("=" * 50)
        
        # Compter les méthodes
        methodes_count = {}
        for colonne, methode in self.methode_utilisee.items():
            methode_base = "Z-score" if "Z-score" in methode else "IQR"
            methodes_count[methode_base] = methodes_count.get(methode_base, 0) + 1
        
        for methode, count in methodes_count.items():
            print(f"{methode}: {count} colonnes")
        
        # Afficher les détails par méthode
        print(f"\n🎯 DÉTAIL PAR MÉTHODE:")
        for methode_base in ["Z-score", "IQR"]:
            colonnes_methode = [col for col, meth in self.methode_utilisee.items() if methode_base in meth]
            if colonnes_methode:
                print(f"\n{methode_base}:")
                for colonne in colonnes_methode[:10]:  # Afficher les 10 premières
                    info = self.outliers_info[colonne]
                    print(f"  {colonne:20} | {info['outliers_detectes']:3} outliers ({info['pourcentage']:.1f}%)")

# Application du code
print("🚀 DÉMARRAGE DU TRAITEMENT DES OUTLIERS")
print("=" * 70)

# Initialisation
treatment = OutlierTreatment(alpha=0.05)

# Traitement des outliers
df_sans_outliers = treatment.traiter_outliers_dataset(df, colonnes_attributs)

# Imputation K-NN
df_final = treatment.imputer_knn(df_sans_outliers, n_neighbors=5)

print("\n🎉 TRAITEMENT TERMINÉ AVEC SUCCÈS!")

```

```python
# Analyse comparative avant/après traitement par le pipeline

def analyser_resultats(df_original, df_final, treatment):
    """Analyse les résultats du traitement"""
    print("\n📈 ANALYSE COMPARATIVE AVANT/APRÈS")
    print("=" * 60)
    
    results = []
    
    for colonne in treatment.colonnes_traitees:
        info = treatment.outliers_info[colonne]
        
        # Statistiques avant
        data_avant = df_original[colonne].dropna()
        # Statistiques après
        data_apres = df_final[colonne]
        
        results.append({
            'Colonne': colonne,
            'Outliers': info['outliers_detectes'],
            '% Outliers': info['pourcentage'],
            'Moyenne Avant': data_avant.mean(),
            'Moyenne Après': data_apres.mean(),
            'Écart-type Avant': data_avant.std(),
            'Écart-type Après': data_apres.std(),
            'Skewness Avant': skew(data_avant),
            'Skewness Après': skew(data_apres)
        })
    
    df_comparison = pd.DataFrame(results)
    
    # Afficher les colonnes avec le plus de changement
    df_comparison['Diff Moyenne'] = abs(df_comparison['Moyenne Après'] - df_comparison['Moyenne Avant'])
    df_comparison['Diff Skewness'] = abs(df_comparison['Skewness Après'] - df_comparison['Skewness Avant'])
    
    print("Top 15 colonnes avec le plus de changement:")
    print(df_comparison.nlargest(15, 'Diff Moyenne')[['Colonne', 'Outliers', 'Moyenne Avant', 'Moyenne Après', 'Diff Moyenne']].to_string(index=False))
    
    return df_comparison

# Appliquer l'analyse
df_comparison = analyser_resultats(df, df_final, treatment)
```

```python
# Visualisation des résultats du traitement

def visualiser_traitement(df_original, df_final, treatment, n_colonnes=6):
    """Visualise l'impact du traitement sur quelques colonnes"""
    colonnes_a_visualiser = treatment.colonnes_traitees[:n_colonnes]
    
    n_rows = len(colonnes_a_visualiser)
    fig, axes = plt.subplots(n_rows, 2, figsize=(12, 3 * n_rows))
    
    if n_rows == 1:
        axes = axes.reshape(1, -1)
    
    for i, colonne in enumerate(colonnes_a_visualiser):
        # Avant traitement
        sns.boxplot(data=df_original, x=colonne, ax=axes[i, 0], color='lightcoral')
        axes[i, 0].set_title(f'AVANT - {colonne}')
        
        # Après traitement
        sns.boxplot(data=df_final, x=colonne, ax=axes[i, 1], color='lightgreen')
        axes[i, 1].set_title(f'APRÈS - {colonne}')
    
    plt.tight_layout()
    plt.suptitle('Comparaison des distributions avant/après traitement des outliers', y=1.02)
    plt.show()

# Visualiser les résultats
visualiser_traitement(df, df_final, treatment, n_colonnes=10)
```

```python
# Définition d'une méthode rapide pour tester la normalité d'une colonne

def test_normalite_grand_echantillon(data, alpha=0.05):
    """Test de normalité optimisé pour grands échantillons"""
    data_clean = pd.Series(data).dropna()
    n = len(data_clean)
    
    if n < 3:
        return False, "Données insuffisantes"
    
    print(f"🔍 Analyse de normalité (n={n})")
    
    # Pour grands échantillons, prioriser d'autres tests
    if n > 5000:
        print("⚠️  Grand échantillon détecté - utilisation de méthodes alternatives")
        
        # 1. Test D'Agostino (meilleur pour grands échantillons)
        try:
            stat_dago, p_dago = normaltest(data_clean)
            normal_dago = p_dago > alpha
            print(f"D'Agostino: p-value = {p_dago:.4f} → {'NORMAL' if normal_dago else 'NON NORMAL'}")
        except Exception as e:
            print(f"D'Agostino: Erreur - {e}")
            normal_dago = False
        
        # 2. Analyse de skewness
        skewness_val = skew(data_clean)
        normal_skew = abs(skewness_val) < 0.5  # Critère plus strict pour grands échantillons
        print(f"Skewness: {skewness_val:.3f} → {'NORMAL' if normal_skew else 'NON NORMAL'}")
        
        # Décision basée sur les deux tests
        est_normal = normal_dago and normal_skew
        
    else:
        # Pour petits échantillons, utiliser Shapiro-Wilk
        try:
            _, p_shapiro = shapiro(data_clean)
            est_normal = p_shapiro > alpha
            print(f"Shapiro-Wilk: p-value = {p_shapiro:.4f} → {'NORMAL' if est_normal else 'NON NORMAL'}")
        except:
            skewness_val = skew(data_clean)
            est_normal = abs(skewness_val) < 1
            print(f"Skewness (fallback): {skewness_val:.3f} → {'NORMAL' if est_normal else 'NON NORMAL'}")
    
    return est_normal
```

```python
# Test de la normaliété de 'int_corr'

resultat = test_normalite_grand_echantillon(df['int_corr'])
print(f"\n🎯 CONCLUSION: La variable est {'NORMALE, donc utilisation du z-score' if resultat else 'NON NORMALE, donc utilisation des IQR'} pour la détection des outliers.")
```

```python
# Définition d'une fonction pour analyser les colonnes d'intérêt :
def describe_column(df, col_name, check_bounds=None, show_value_counts=False):

   # Paramètres:
    # check_bounds : Si spécifié, vérifie les valeurs en dehors des bornes [lower, upper]
    # show_value_counts : Si True, affiche la répartition des valeurs uniques


    # 1. Affichage du type de données
    print(f"{col_name} : {df[col_name].dtype}")
    
    # 2. Nombre de valeurs uniques
    print(f"Nombre de valeurs uniques : {df[col_name].nunique()}")
    
    # 3. Nombre de valeurs manquantes (NaN)
    print(f"Nombre de valeurs manquantes : {df[col_name].isna().sum()}")
    
    # 4. Vérification des valeurs aberrantes (si check_bounds est spécifié)
    if check_bounds :
        lower, upper = check_bounds  # Dépaquetage des bornes
        print(f"Nombre de valeurs supérieures à {upper} : {(df[col_name] > upper).sum()}")
        print(f"Nombre de valeurs inférieures à {lower} : {(df[col_name] < lower).sum()}")
    
    # 5. Affichage de la répartition des valeurs (si demandé)
    if show_value_counts:
        print("Répartition des valeurs :")
        print(df[col_name].value_counts())
    
    print(30 * '-')

```
